```python
import numpy as np
import matplotlib.pyplot as plt


gamma = 1.0   
n = 2.0       
sigma = 0.1   
dt = 0.001     
T = 10        
N = int(T / dt) 

x = np.zeros(N)
y = np.zeros(N)
x[0], y[0] = 1.0, 0.5 

dW_x = np.random.normal(0, np.sqrt(dt), N)
dW_y = np.random.normal(0, np.sqrt(dt), N)

for t in range(N - 1):
    dx = gamma * (y[t] - x[t]) * dt + sigma * dW_x[t]
    dy = (2 / (1 + x[t]**n) - y[t]) * dt + sigma * dW_y[t]
    x[t + 1] = x[t] + dx
    y[t + 1] = y[t] + dy

plt.figure(figsize=(10, 5))
plt.plot(np.linspace(0, T, N), x, label="x(t)", color="b")
plt.plot(np.linspace(0, T, N), y, label="y(t)", color="r")
plt.xlabel("Time t")
plt.ylabel("x, y")
plt.legend()
plt.title("Stochastic Dynamics with Euler-Maruyama Method")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
V = 120
k = 10
gamma = 2
n = 10
sigma = 5

# Time parameters
T = 100
dt = 0.01
steps = int(T / dt)

# Initial condition
x = np.zeros(steps)
x[0] = 10  # Start from a large x

# Euler-Maruyama simulation
for i in range(1, steps):
    production = V * (x[i-1]**n) / (k**n + x[i-1]**n)
    degradation = gamma * x[i-1]
    dx = (production - degradation) * dt + sigma * x[i-1] * np.sqrt(dt) * np.random.normal()
    x[i] = x[i-1] + dx

# Plot results
t = np.linspace(0, T, steps)
plt.plot(t, x, label=f'σ = {sigma}')
plt.xlabel('Time')
plt.ylabel('x(t)')
plt.legend()
plt.show()
```
