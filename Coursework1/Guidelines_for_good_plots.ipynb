In your coursework, you need to produce some plots. Here, we outline how a "good" plot looks like.

For this purpose, let's consider a specific example of data belonging to different classes, and we generate it using sklearn's [`make_classification`](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_classification.html) function.

If we just plot the data using the standard settings, we don't end up with a very good plot.

```python
# We create a dataset of 500 data points, divided into 2 classes, each data point is described by 2 features
from sklearn.datasets import make_classification
import numpy as np
import matplotlib.pyplot as plt
X, y = make_classification(n_samples=500, n_features=2, n_redundant=0, n_informative=2, n_clusters_per_class=1, n_classes=2, random_state=15)

# we plot the data points on the plane, marking the class with the colour.
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.show()
```

As a rule of thumb, **a plot should always be able to stand alone**: it should be possible to understand the plot without reading the surrounding code or text, and even (if possible) without reading the figure caption.

The plot above can clearly not stand on its own: it doesn't have any legend, nor does it have labels on the axes. Likewise, we should add a title. Other types of plots might have other options to add information. For instance, if we were dealing with a heatmap, we need to include a colour bar to indicate the quantitative scale for which colours stand for.

```python
labels = ["Class 1", "Class 2"]
for i, c in enumerate(np.unique(y)):
    plt.scatter(X[:, 0][y==c], X[:, 1][y==c], label=labels[i]);

plt.xlabel("Feature 1");
plt.ylabel("Feature 2");

plt.title("Clustering of data into 2 classes");
plt.legend();
```

These small additions have improved the plot significantly. But it's still not optimal. First of all, it's rather small. To make it more readable, we should increase the figure size and change the font size of the different labels, titles, and ticks. Now, what is a suitable font size? As a rule of thumb, **a plot should be easy to read when you print it**, which is the case when its ticks/labels/titles etc should have size at least comparable to the surrounding text.

A convenient strategy is to reset the default parameters for fontsizes in matplotlib, in such a way as to avoid having to set these parameters for each plot. We recommend adding a cell at the beginning of your notebooks setting these parameter values, as you have seen in our coding tasks.


```python
# we define a mix of fontsizes, for different parts of a plot
SMALL_SIZE = 12
MEDIUM_SIZE = 16
BIGGER_SIZE = 20

# example of how you can use these fontsizes to set a global configuration for matplotlib;
# you should assign them based on the specific appearance of the figures you are producing
plt.rc('font', size=SMALL_SIZE)          # controls default text sizes
plt.rc('axes', titlesize=BIGGER_SIZE)    # fontsize of the axes title
plt.rc('axes', labelsize=MEDIUM_SIZE)    # fontsize of the x and y labels
plt.rc('xtick', labelsize=SMALL_SIZE)    # fontsize of the tick labels
plt.rc('ytick', labelsize=SMALL_SIZE)    # fontsize of the tick labels
plt.rc('legend', fontsize=MEDIUM_SIZE)   # legend fontsize
plt.rc('figure', titlesize=BIGGER_SIZE)  # fontsize of the figure title


fig, ax = plt.subplots(figsize=(10,6)) # We make the plot larger

labels = ["Class 1", "Class 2"]
for i, c in enumerate(np.unique(y)):
    plt.scatter(X[:, 0][y==c], X[:, 1][y==c], label=labels[i]);

plt.xlabel("Feature 1");
plt.ylabel("Feature 2");

plt.title("Clustering of data into 2 classes");
ax.legend(loc="best");
```

The plot above is what we would consider a "good" plot since it fulfils both the criteria sketched above: it can stand alone, and it's easy to read. But what if we had decided to plot something different, such as the strengh of a penalty term, i.e. lambda? In that case, LaTeX commands would be useful for writing $\lambda$ rather than "lambda". Indeed, we can readily do so in Python by passing a raw string (r"") with a LaTeX command. So, for the sake of the example, let's say that we have called feature 1 $x_{\alpha}$ and the feature 2 $x_{\beta}$.

```python
fig, ax = plt.subplots(figsize=(10,6)) # We make plot larger

labels = ["Class 1", "Class 2"]
for i, c in enumerate(np.unique(y)):
    plt.scatter(X[:, 0][y==c], X[:, 1][y==c], label=labels[i]);

plt.xlabel(r"$x_{\alpha}$"); ## Now here we use latex commands
plt.ylabel(r"$x_{\beta}$");

plt.title("Clustering of data into 2 classes");
ax.legend(loc="best");
```

[Extra considerations -- not needed for the coursework] We might want to reconsider the markers as well. Indeed, it is good practice to produce plots that should be easy to read when you print them **in grayscale**, that is, you should choose [colormaps](https://matplotlib.org/stable/tutorials/colors/colormaps.html) that are grayscale convertible, which is important also to make the plot colourblind-friendly. In general, check that the colourmap you choose is colour-blind-friendly, `viridis` is a good example. The default colormap of plt.scatter is actually grayscale convertible (to see what grayscale convertible means, you can try to set the grayscale using [plt.style.use('grayscale')](https://matplotlib.org/stable/gallery/style_sheets/grayscale.html)). If in doubt, you can, of course, use your own colour scheme.

Another way to make the plot more readable when printed in grayscale (and hence more readable in general) is to use different markers in addition to colours that are easily distinguishable. We can, therefore, also change the markers themselves below. You can use the same trick when you make other plots. If you were to make, say, a line plot, you could consider using different line styles ("-",":","--").

```python
ig, ax = plt.subplots(figsize=(10,6)) # We make plot larger

labels = ["Class 1", "Class 2"]
markers = ["^","o"] # We use distinct markers
for i, c in enumerate(np.unique(y)):
    plt.scatter(X[:, 0][y==c], X[:, 1][y==c], marker=markers[i], label=labels[i]);

plt.xlabel(r"$x_{\alpha}$"); ## Now here we use latex commands
plt.ylabel(r"$x_{\beta}$");

plt.title("Clustering of data into 2 classes");
ax.legend(loc="best");
```
